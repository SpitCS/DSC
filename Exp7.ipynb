```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set Seaborn theme
sns.set_theme(style="whitegrid")

# Load Titanic dataset from Seaborn
df = sns.load_dataset('titanic')

# Display first five rows
display(df.head())

# Display basic information
df.info()

# Display quick statistics
display(df.describe(include='all'))
```

```python
# 1. Plot a histogram of a numeric feature using Matplotlib.

age_data = df['age'].dropna()
plt.figure(figsize=(8, 5))
plt.hist(age_data, bins=10)
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# 2. Create a bar chart to show counts of each category in a column.

class_counts = df['class'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(class_counts.index, class_counts.values)
plt.title("Number of Passengers by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
# 3. Use seaborn.countplot() to visualize category frequency.

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='class')
plt.title("Passenger Count by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
# 4. Draw a Seaborn histplot for a single variable.

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='age', bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# 5. Create a boxplot to visualize distribution and detect outliers.

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='age')
plt.title("Boxplot of Age")
plt.xlabel("Age")
plt.show()
```

```python
# 6. Plot a pie chart showing percentage distribution of categories.

class_counts = df['class'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    class_counts.values,
    labels=class_counts.index,
    autopct='%1.1f%%'
)
plt.title("Percentage Distribution of Passenger Classes")
plt.show()
```

```python
# 7. Use seaborn.violinplot() to show spread of a numeric feature.

plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='age')
plt.title("Age Distribution Using Violin Plot")
plt.xlabel("Age")
plt.show()
```

```python
# 8. Visualize the distribution of a numeric column with KDE.

age_data = df['age'].dropna()

plt.figure(figsize=(8, 5))
sns.kdeplot(age_data)
plt.title("Kernel Density Estimate of Age")
plt.xlabel("Age")
plt.ylabel("Density")
plt.show()
```

```python
# 9. Draw side-by-side histograms comparing two numeric features.

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(df['age'].dropna(), bins=10)
axes[0].set_title("Age Distribution")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Frequency")

axes[1].hist(df['fare'].dropna(), bins=10)
axes[1].set_title("Fare Distribution")
axes[1].set_xlabel("Fare")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# 10. Customize histogram with bins, color, and alpha transparency.

age_data = df['age'].dropna()

plt.figure(figsize=(8, 5))
plt.hist(
    age_data,
    bins=15,
    color='skyblue',
    alpha=0.7
)
plt.title("Customized Age Histogram")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# 11. Create subplots of univariate distributions for multiple features.

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].hist(df['age'].dropna(), bins=10)
axes[0].set_title("Age")

axes[1].hist(df['fare'].dropna(), bins=10)
axes[1].set_title("Fare")

axes[2].hist(df['sibsp'].dropna(), bins=10)
axes[2].set_title("Siblings/Spouses")

plt.tight_layout()
plt.show()
```

```python
# 12. Use value_counts().plot(kind='bar') on a categorical column.

df['class'].value_counts().plot(kind='bar', figsize=(8, 5))

plt.title("Passenger Count by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
# 13. Create a dot plot (strip plot) using Seaborn for univariate data.

age_data = df[['age']].dropna()

plt.figure(figsize=(8, 3))
sns.stripplot(data=age_data, x='age')
plt.title("Strip Plot of Age")
plt.xlabel("Age")
plt.show()
```

```python
# 14. Plot a line chart of sorted values of a single numeric feature.

sorted_age = df['age'].dropna().sort_values().reset_index(drop=True)

plt.figure(figsize=(10, 5))
plt.plot(sorted_age)
plt.title("Sorted Age Values")
plt.xlabel("Index")
plt.ylabel("Age")
plt.show()
```

```python
# 15. Use sns.ecdfplot() to draw an empirical cumulative distribution plot.

plt.figure(figsize=(8, 5))
sns.ecdfplot(data=df, x='age')
plt.title("Empirical Cumulative Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Cumulative Probability")
plt.show()
```

```python
# 16. Plot a lollipop chart for ordinal categorical data.

class_counts = df['class'].value_counts().sort_index()

x = range(len(class_counts))
y = class_counts.values

plt.figure(figsize=(8, 5))
plt.stem(x, y)

plt.xticks(x, class_counts.index)
plt.title("Lollipop Chart of Passenger Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
# 17. Show percentage labels on top of bars in a bar plot.

class_counts = df['class'].value_counts()
total = class_counts.sum()

plt.figure(figsize=(8, 5))
bars = plt.bar(class_counts.index, class_counts.values)

for bar, value in zip(bars, class_counts.values):
    percentage = value / total * 100
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{percentage:.1f}%',
        ha='center',
        va='bottom'
    )

plt.title("Passenger Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
# 18. Rotate x-axis labels for readability in a bar chart.

class_counts = df['class'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(class_counts.index, class_counts.values)

plt.title("Passenger Count by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# 19. Display mean, median, and mode lines on a histogram.

age_data = df['age'].dropna()

mean_age = age_data.mean()
median_age = age_data.median()
mode_age = age_data.mode()[0]

plt.figure(figsize=(9, 5))
plt.hist(age_data, bins=15, alpha=0.7)

plt.axvline(mean_age, linestyle='--', label=f'Mean = {mean_age:.2f}')
plt.axvline(median_age, linestyle='-.', label=f'Median = {median_age:.2f}')
plt.axvline(mode_age, linestyle=':', label=f'Mode = {mode_age:.2f}')

plt.title("Age Distribution with Mean, Median and Mode")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# 30. Annotate mean and standard deviation on a boxplot.

age_data = df['age'].dropna()

mean_age = age_data.mean()
std_age = age_data.std()

plt.figure(figsize=(8, 5))

sns.boxplot(x=age_data)

plt.text(
    mean_age,
    0.1,
    f"Mean = {mean_age:.2f}\nStd = {std_age:.2f}",
    ha='center'
)

plt.title("Age Boxplot with Mean and Standard Deviation")
plt.xlabel("Age")
plt.show()
```

```python
# 31. Compare value counts from two datasets in one bar plot.

dataset1 = df['class'].value_counts().sort_index()

# Create a second dataset by using the male passengers
dataset2 = df[df['sex'] == 'male']['class'].value_counts().sort_index()

comparison = pd.DataFrame({
    'All Passengers': dataset1,
    'Male Passengers': dataset2
}).fillna(0)

comparison.plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title("Comparison of Passenger Class Counts")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```
